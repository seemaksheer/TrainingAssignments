```python
#===============================================================================
# Program Name : Employee_Attrition_KNN_Classification.ipynb
# Purpose      : Predict whether an employee will leave the organization
#                 using the K-Nearest Neighbors (KNN) Classification Algorithm.
#
# Machine Learning Type : Supervised Learning
# Algorithm             : K-Nearest Neighbors (KNN) Classifier
#
# Target Variable :
#       Attrition
#       1 = Employee Left the Company
#       0 = Employee Stayed in the Company
```

```python
# STEP 1 - Import Required Libraries
#
# Pandas is used for handling datasets in tabular format.
import pandas as pd

# StringIO helps convert text into a CSV file format.
from io import StringIO

# train_test_split is used to divide the dataset into Training and Testing data.
from sklearn.model_selection import train_test_split

# LabelEncoder converts text values into numeric values.
from sklearn.preprocessing import LabelEncoder

# StandardScaler standardizes numerical values.
from sklearn.preprocessing import StandardScaler

# KNeighborsClassifier is the KNN Machine Learning algorithm.
from sklearn.neighbors import KNeighborsClassifier

# Accuracy Score measures prediction accuracy.
from sklearn.metrics import accuracy_score
print("=" * 80)
print("Required Libraries Loaded Successfully")
print("=" * 80)
```

```python
# STEP 2 - Create the Dataset
# Instead of reading from an external CSV file,
# Create the dataset directly as a multi-line string.
data = """Age,JobRole,MonthlyIncome,JobSatisfaction,YearsAtCompany,Attrition
29,Sales Executive,4800,3,4,1
35,Research Scientist,6000,4,8,0
40,Laboratory Technician,3400,2,6,0
28,Sales Executive,4300,3,3,1
45,Manager,11000,4,15,0
25,Research Scientist,3500,1,2,1
50,Manager,12000,4,20,0
30,Sales Executive,5000,2,5,0
37,Laboratory Technician,3100,2,9,0
26,Research Scientist,4500,3,2,1
"""

# Read the CSV text into a Pandas DataFrame.
df = pd.read_csv(StringIO(data))

# Display the complete dataset.
print("=" * 80)
print("Employee Dataset")
print("=" * 80)
print(df)
```

```python
# STEP 3 - Convert Categorical Data into Numeric Data
# Text converted to numbers
# Convert JobRole into numbers.
encoder = LabelEncoder()
df["JobRole"] = encoder.fit_transform(df["JobRole"])
print("\n")
print("=" * 80)
print("Dataset after Label Encoding")
print("=" * 80)
print(df)
print("\nEncoded Job Roles")
print(dict(zip(encoder.classes_, encoder.transform(encoder.classes_))))
```

```python
# STEP 4 - Separate Features (X) and Target Variable (y)
#===============================================================================
# X contains all input columns.
X = df.drop("Attrition", axis=1)
# y contains only the target column.
y = df["Attrition"]
print("\n")
print("=" * 80)
print("Features (X)")
print("=" * 80)
print(X)
print("\nTarget (y)")
print(y)
```

```python
#===============================================================================
# STEP 5 - Split Dataset into Training and Testing Sets
# 80% data is used for training.
# 20% data is used for testing.

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
print("\n")
print("=" * 80)
print("Training Data Shape")
print(X_train.shape)
print("Testing Data Shape")
print(X_test.shape)
```

```python
#===============================================================================
# STEP 6 - Feature Scaling
# ==============================================================================
# KNN works based on distance.
# Numerical values should be on the same scale.
scaler = StandardScaler()
# Learn scaling from training data.
X_train = scaler.fit_transform(X_train)
# Apply the same scaling to testing data.
X_test = scaler.transform(X_test)
```

```python
#===============================================================================
# STEP 7 - Train the KNN Classification Model
#===============================================================================
# Create KNN model.
# n_neighbors=3 means consider the nearest 3 employees.
model = KNeighborsClassifier(n_neighbors=3)

# Train the model.

model.fit(X_train, y_train)

print("\n")
print("=" * 80)
print("Model Training Completed Successfully")
print("=" * 80)
```

```python
#===============================================================================
# STEP 8 - Predict Attrition for Test Data
#===============================================================================
predictions = model.predict(X_test)

print("\nActual Values")
print(y_test.values)

print("\nPredicted Values")
print(predictions)
```

```python
#===============================================================================
# STEP 9 - Calculate Model Accuracy
#===============================================================================
accuracy = accuracy_score(y_test, predictions)
print("\n")
print("=" * 80)
print(f"Model Accuracy : {accuracy:.2f}")
print("=" * 80)
```

```python
#===============================================================================
# STEP 10 - Predict Attrition for a New Employee
#===============================================================================
# New Employee Details
#
# Age               = 27
# JobRole           = Research Scientist
# Monthly Income    = 4200
# Job Satisfaction  = 2
# Years At Company  = 0

print("\n")
print("=" * 100)
print("\nSCENARIO 1 : Employee Likely to Leave the Company")
print("-" * 100)

new_employee = pd.DataFrame({
    "Age": [27],
    "JobRole": ["Research Scientist"],
    "MonthlyIncome": [4200],
    "JobSatisfaction": [2],
    "YearsAtCompany": [0]
})

# Convert JobRole into numeric value.
new_employee["JobRole"] = encoder.transform(new_employee["JobRole"])

# Scale the values.
new_employee_scaled = scaler.transform(new_employee)
print(new_employee_scaled)

# Predict Attrition.
prediction = model.predict(new_employee_scaled)

print("\n")
print("=" * 80)
print("Prediction for New Employee Likely to leave the Company")
print("=" * 80)

print("\nPrediction Value :", prediction[0])

if prediction[0] == 1:
    print("Prediction : Employee is likely to Leave the Company.")
else:
    print("Prediction : Employee is likely to Stay in the Company.")
```

```python
#===============================================================================
# STEP 10 - Predict Attrition for a New Employee Likely to Stay in the Company
#===============================================================================
# New Employee Details
#===============================================================================

print("\n")
print("=" * 100)
print("SCENARIO 2 : Employee Likely to Continue to Stay in the Company")
print("-" * 100)

employee_stay = pd.DataFrame({
    "Age": [46],
    "JobRole": ["Manager"],
    "MonthlyIncome": [11500],
    "JobSatisfaction": [4],
    "YearsAtCompany": [16]
})

print("\nEmployee Details")
print(employee_stay)

# Convert JobRole into numeric value
employee_stay["JobRole"] = encoder.transform(employee_stay["JobRole"])

print("\nAfter Label Encoding")
print(employee_stay)

# Scale the data
employee_stay_scaled = scaler.transform(employee_stay)

print("\nScaled Employee Data")
print(employee_stay_scaled)

# Predict
prediction = model.predict(employee_stay_scaled)
print("\n")
print("=" * 80)
print("Prediction for New Employee Likely to Continue to Stay in the Company")
print("=" * 80)

print("\nPrediction Value :", prediction[0])

if prediction[0] == 1:
    print("Prediction : Employee is likely to Leave the Company.")
else:
    print("Prediction : Employee is likely to Continue to Stay in the Company.")
```
