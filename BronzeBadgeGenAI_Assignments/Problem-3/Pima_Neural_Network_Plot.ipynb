```python
# ==============================================================================
# Program Name : Pima_Indians_Diabetes_Neural_Network_Visualization.py
#
# Description:
# This program visualizes the architecture of a Feed Forward Neural Network
# used for the Pima Indians Diabetes Prediction problem.
#
# Neural Network Architecture
# ---------------------------
# Input Layer    : 8 Neurons (Dataset Features)
# Hidden Layer 1 : 12 Neurons
# Hidden Layer 2 : 8 Neurons
# Output Layer   : 1 Neuron (Diabetes Prediction)
#
# Developed Using:
#   - Python
#   - Matplotlib
# ==============================================================================
```

```python
# Import the matplotlib library for drawing the neural network.
import matplotlib.pyplot as plt
```

```python
# ==============================================================================
# Function: draw_neural_network()
#
# Purpose:
# Draws the neural network architecture based on the number of neurons
# specified in each layer.
#
# Parameters:
#   ax          -> Matplotlib axis object
#   layer_sizes -> List containing the number of neurons in each layer
#                  Example: [8,12,8,1]
# ==============================================================================
def draw_neural_network(ax, layer_sizes):

    # Horizontal distance between two layers.
    h_spacing = 2

    # Vertical distance between neurons.
    v_spacing = 1

    # Find the layer with the maximum number of neurons.
    max_neurons = max(layer_sizes)

    # Store the coordinates of every neuron.
    neuron_positions = []

    # --------------------------------------------------------------------------
    # Draw all neurons layer by layer.
    # --------------------------------------------------------------------------
    for layer, neurons in enumerate(layer_sizes):

        # X-coordinate of the current layer.
        layer_x = layer * h_spacing

        # Center the neurons vertically.
        top = (max_neurons - neurons) / 2

        current_layer = []

        # Draw each neuron as a circle.
        for neuron in range(neurons):

            layer_y = top + neuron

            circle = plt.Circle(
                (layer_x, layer_y),   # Position
                0.15,                 # Radius
                color="skyblue",      # Fill Color
                ec="black"            # Border Color
            )

            ax.add_artist(circle)

            current_layer.append((layer_x, layer_y))

        neuron_positions.append(current_layer)

    # --------------------------------------------------------------------------
    # Draw the connections between adjacent layers.
    # Every neuron is connected to every neuron in the next layer.
    # --------------------------------------------------------------------------
    for layer in range(len(layer_sizes)-1):

        for start in neuron_positions[layer]:

            for end in neuron_positions[layer+1]:

                ax.plot(
                    [start[0], end[0]],
                    [start[1], end[1]],
                    color="gray",
                    linewidth=0.5
                )

    # --------------------------------------------------------------------------
    # Add labels above every layer.
    # --------------------------------------------------------------------------
    ax.text(
        0,
        max_neurons + 0.5,
        "Input\n(8 Features)",
        ha="center",
        fontsize=11,
        weight="bold"
    )

    ax.text(
        2,
        max_neurons + 0.5,
        "Hidden\nLayer 1\n(12)",
        ha="center",
        fontsize=11,
        weight="bold"
    )

    ax.text(
        4,
        max_neurons + 0.5,
        "Hidden\nLayer 2\n(8)",
        ha="center",
        fontsize=11,
        weight="bold"
    )

    ax.text(
        6,
        max_neurons + 0.5,
        "Output\n(1)",
        ha="center",
        fontsize=11,
        weight="bold"
    )

    # Set the figure limits.
    ax.set_xlim(-1, 7)
    ax.set_ylim(-1, max_neurons + 2)

    # Keep neurons circular instead of oval.
    ax.set_aspect("equal")

    # Remove axis.
    ax.axis("off")


# ==============================================================================
# Main Program
# ==============================================================================

# Create a figure.
fig = plt.figure(figsize=(12, 8))

# Create an axis.
ax = fig.gca()

# Specify the neural network architecture.
# 8 Input -> 12 Hidden -> 8 Hidden -> 1 Output
layer_sizes = [8, 12, 8, 1]

# Draw the neural network.
draw_neural_network(ax, layer_sizes)

# Add the title.
plt.title(
    "Neural Network for Pima Indians Diabetes Prediction",
    fontsize=16,
    weight="bold"
)

# Display the diagram.
plt.show()
```
