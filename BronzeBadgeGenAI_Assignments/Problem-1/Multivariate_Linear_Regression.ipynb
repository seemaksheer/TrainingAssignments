```python
# =============================================================================
# Problem Statement
# =============================================================================
#
# Build a Multivariate Linear Regression model to predict product sales
# based on the advertising budget allocated to different marketing channels.
#
# Input Features (Independent Variables):
#   1. TV Budget ($)
#   2. Radio Budget ($)
#   3. Newspaper Budget ($)
#
# Output (Dependent Variable):
#   Sales (units)
#
# Objective:
# Train a Multivariate Linear Regression model using the given dataset
# so that it can learn the relationship between the advertising budgets
# and product sales. After training, use the model to predict the sales
# for a new advertising budget.
#
# Machine Learning Technique:
# Multivariate Linear Regression
#
# =============================================================================
```

```python
# =============================================================================
# STEP 1 - Import Required Libraries
# =============================================================================
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
# =============================================================================
# STEP 2 - Create the Dataset
# =============================================================================
data = {
    "TV": [230.1,44.5,17.2,151.5,180.8,8.7,57.5,120.2,144.1,111.6],
    "Radio": [37.8,39.3,45.9,41.3,10.8,48.9,32.8,19.6,16.0,12.6],
    "Newspaper": [69.2,45.1,69.3,58.5,58.4,75.0,23.5,11.6,40.3,37.9],
    "Sales": [22.1,10.4,9.3,18.5,12.9,7.2,11.8,13.2,15.6,12.2]
}
```

```python
# =============================================================================
# STEP 3 - Convert Dictionary into DataFrame
# =============================================================================
df = pd.DataFrame(data)
print(df)
```

```python
# =============================================================================
# STEP 4 - Select the Input Features
# =============================================================================
X = df[["TV", "Radio", "Newspaper"]]
print("\nInput Features")
print(X)
```

```python
# =============================================================================
# STEP 5 - Select the Target Variable
# =============================================================================
y = df["Sales"]
print("\nTarget Variable")
print(y)
```

```python
# =============================================================================
# STEP 6 - Create the Linear Regression Model
# =============================================================================
model = LinearRegression()
```

```python
# =============================================================================
# STEP 7 - Train the Model
# =============================================================================
model.fit(X, y)
```

```python
# =============================================================================
# STEP 8 - Display the Regression Equation
# =============================================================================
print("Intercept:", model.intercept_)
print("\nCoefficients")
print("TV Budget       :", model.coef_[0])
print("Radio Budget    :", model.coef_[1])
print("Newspaper Budget:", model.coef_[2])
```

```python
# =============================================================================
# STEP 9 - Predict Sales for a New Advertising Budget
# =============================================================================
new_budget = pd.DataFrame({
    "TV": [100],
    "Radio": [25],
    "Newspaper": [40]
})

predicted_sales = model.predict(new_budget)

print("\nPredicted Sales")
print("TV Budget        : $100")
print("Radio Budget     : $25")
print("Newspaper Budget : $40")
print("Predicted Sales  :", predicted_sales[0])
```
